### 1.3 Exceptions

Write `safe_divide(a, b)` that:
- Returns `a / b` normally
- Returns `None` if `b == 0` (catch `ZeroDivisionError`)
- Raises `TypeError` with message `"Inputs must be numeric"` if either input is not a number

````python
def safe_divide(a, b):
    if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
        raise TypeError("Inputs must be numeric")
    try:
        return a / b
    except ZeroDivisionError:
        return None

# Tests
assert safe_divide(10, 2) == 5.0
assert safe_divide(5, 0) is None
try:
    safe_divide('x', 2)
    assert False, "Should have raised TypeError"
except TypeError as e:
    assert str(e) == "Inputs must be numeric"
print("1.3 passed ✓")
````

### 4.3 Eigenvalues & Eigenvectors

For the matrix:
```
A = [[4, 1],
     [2, 3]]
```

1. Compute eigenvalues and eigenvectors using `np.linalg.eig`.
2. **Verify** the eigenvector equation: `A @ v = λ * v` for each eigenpair.
3. Plot both eigenvectors as arrows, scaled by their eigenvalues, on a 2D plot.
4. **Explain** in a markdown cell (below): what does it geometrically mean for a matrix to stretch a vector?

````python
A = np.array([[4, 1], [2, 3]], dtype=float)

eigenvalues, eigenvectors = np.linalg.eig(A)

print(f"Eigenvalues: {eigenvalues}")
print(f"Eigenvectors (columns):\n{eigenvectors}")

# Verify Av = λv for each eigenpair
for i in range(len(eigenvalues)):
    lam = eigenvalues[i]
    v   = eigenvectors[:, i]
    Av  = A @ v
    lv  = lam * v
    match = np.allclose(Av, lv)
    print(f"  λ={lam:.4f}: A@v={Av.round(4)}, λ*v={lv.round(4)}, match={match}")

# Plot eigenvectors scaled by eigenvalues
fig, ax = plt.subplots()
colors = ['steelblue', 'darkorange']
for i in range(len(eigenvalues)):
    vec = eigenvectors[:, i] * eigenvalues[i]
    ax.quiver(0, 0, vec[0], vec[1], angles='xy', scale_units='xy', scale=1,
              color=colors[i], width=0.015, label=f'λ={eigenvalues[i]:.1f}')
ax.set_xlim(-4, 4); ax.set_ylim(-4, 4)
ax.axhline(0, color='k', lw=0.8); ax.axvline(0, color='k', lw=0.8)
ax.legend(); ax.set_title('Eigenvectors scaled by Eigenvalues')
plt.tight_layout(); plt.savefig('/tmp/eig.png', dpi=60); plt.close()

print("4.3 passed ✓")
````

**Your Geometric Explanation (4.3):**

When a matrix **A** is applied to a vector, it generally rotates and scales the vector, changing both its direction and length. However, **eigenvectors** are special directions that a matrix only *scales* — it does not rotate them. The matrix merely stretches or compresses the eigenvector along its own direction by a factor called the **eigenvalue** (λ). If λ > 1, the vector is stretched; if 0 < λ < 1, it is squished; if λ < 0, it is flipped and scaled. This means eigenvectors represent the "natural axes" of a linear transformation — the directions the matrix acts on most simply.

### 4.4 SVD & Dimensionality Reduction

1. Create a random 4×3 matrix `X` (use `np.random.seed(42)`).
2. Compute its **SVD**: `U, S, Vt = np.linalg.svd(X, full_matrices=False)`.
3. **Reconstruct** `X` from the SVD and verify it matches the original.
4. Perform a **rank-1 approximation** using only the largest singular value/vector — call it `X_approx`.
5. Explain the relationship between SVD components and PCA in the markdown cell below.

````python
np.random.seed(42)
X = np.random.randn(4, 3)

U, S, Vt = np.linalg.svd(X, full_matrices=False)

X_reconstructed = U @ np.diag(S) @ Vt
X_approx        = S[0] * np.outer(U[:, 0], Vt[0, :])

print(f"U shape: {U.shape}, S shape: {S.shape}, Vt shape: {Vt.shape}")
print(f"Singular values: {S}")
print(f"Reconstruction error: {np.linalg.norm(X - X_reconstructed):.2e}")
print(f"Rank-1 approximation:\n{X_approx}")

assert np.allclose(X, X_reconstructed, atol=1e-10), "Reconstruction failed!"
print("4.4 passed ✓")
````

**SVD → PCA Connection (4.4):**

In the SVD decomposition **X = U S Vᵀ**:
- **Vᵀ** (right singular vectors) contains the **principal component directions** — the orthogonal axes of maximum variance in the data. Each row of Vᵀ is a principal component direction.
- **S** (singular values) encodes how much variance is captured along each principal component direction. The square of each singular value, divided by (n-1), equals the eigenvalue of the covariance matrix, i.e. the variance along that component.
- **U** (left singular vectors) gives the coordinates of data points in the principal component space (scores).

When you centre the data matrix X and compute its SVD, the columns of V (rows of Vᵀ) are identical to the eigenvectors of the covariance matrix Xᵀ X / (n-1) — the same quantity PCA finds. Thus SVD on the centred data matrix is an efficient way to perform PCA without explicitly computing the covariance matrix.

### 5.4 Distribution Testing & Stationarity

1. Generate two samples: `s1` ~ Normal(0, 1), `s2` ~ Exponential(1), both n=200 (seed=0).
2. Run the **Kolmogorov-Smirnov test** (`stats.kstest`) comparing each against a standard normal CDF.
3. Generate a non-stationary time series (linear trend + noise). Run the **Augmented Dickey-Fuller test** (`adfuller` from `statsmodels`) and interpret the result.
4. Difference the series once and re-run ADF to confirm stationarity.

````python
# statsmodels not available in this environment; using scipy-based implementation
import numpy as np

np.random.seed(0)
s1 = np.random.normal(0, 1, 200)
s2 = np.random.exponential(1, 200)

# KS test — compare each against N(0,1)
ks_s1 = stats.kstest(s1, 'norm', args=(0, 1))
ks_s2 = stats.kstest(s2, 'norm', args=(0, 1))
print(f"KS test s1 (normal):      stat={ks_s1.statistic:.4f}, p={ks_s1.pvalue:.4f}")
print(f"KS test s2 (exponential): stat={ks_s2.statistic:.4f}, p={ks_s2.pvalue:.4f}")

# ADF test — manual implementation using OLS regression
def adfuller_manual(ts):
    """Simplified ADF test (no lag, no trend) using OLS."""
    y = np.array(ts)
    dy = np.diff(y)
    y_lag = y[:-1]
    # OLS: dy = alpha + rho * y_lag
    X = np.column_stack([np.ones(len(y_lag)), y_lag])
    beta, _, _, _ = np.linalg.lstsq(X, dy, rcond=None)
    rho = beta[1]
    y_hat = X @ beta
    resid = dy - y_hat
    s2 = np.sum(resid**2) / (len(dy) - 2)
    var_rho = s2 * np.linalg.inv(X.T @ X)[1, 1]
    t_stat = rho / np.sqrt(var_rho)
    # Approximate p-value (ADF critical values at 5%: ~ -2.86)
    # Use t-distribution as rough proxy
    p_approx = stats.t.cdf(t_stat, df=len(dy)-2)
    return (t_stat, p_approx)

t = np.arange(200)
ts = 0.05 * t + np.random.normal(0, 1, 200)

adf_result = adfuller_manual(ts)
print(f"\nADF statistic: {adf_result[0]:.4f}")
print(f"ADF p-value:   {adf_result[1]:.4f}")
print(f"Stationary?    {adf_result[1] < 0.05}")

# Difference once and re-run
ts_diff  = np.diff(ts)
adf_diff = adfuller_manual(ts_diff)
print(f"\nAfter differencing — p-value: {adf_diff[1]:.4f}, Stationary? {adf_diff[1] < 0.05}")
````

````python
def compute_psi(expected, actual, bins=10):
    """
    PSI = sum((actual_pct - expected_pct) * ln(actual_pct / expected_pct))
    Add epsilon=1e-10 to avoid log(0).
    """
    eps = 1e-10
    breakpoints = np.percentile(expected, np.linspace(0, 100, bins + 1))
    breakpoints[0]  -= eps
    breakpoints[-1] += eps

    exp_counts = np.histogram(expected, bins=breakpoints)[0]
    act_counts = np.histogram(actual,   bins=breakpoints)[0]

    exp_pct = exp_counts / len(expected) + eps
    act_pct = act_counts / len(actual)   + eps

    psi = np.sum((act_pct - exp_pct) * np.log(act_pct / exp_pct))
    return psi

np.random.seed(1)
train_dist = np.random.normal(50, 10, 1000)
drift_dist = np.random.normal(65, 12, 1000)

psi_value = compute_psi(train_dist, drift_dist)
print(f"PSI: {psi_value:.4f}")
print(f"Shift severity: {'Major' if psi_value > 0.2 else 'Minor' if psi_value > 0.1 else 'Stable'}")

# Plot both distributions
from scipy.stats import gaussian_kde
fig, ax = plt.subplots()
for dist, label, color in [(train_dist, 'Training (Expected)', 'steelblue'),
                            (drift_dist, 'Production (Actual)', 'darkorange')]:
    kde = gaussian_kde(dist)
    x_range = np.linspace(20, 110, 300)
    ax.plot(x_range, kde(x_range), label=label, color=color, lw=2)
    ax.fill_between(x_range, kde(x_range), alpha=0.2, color=color)
ax.legend(); ax.set_title('Distribution Shift Visualization')
plt.tight_layout(); plt.savefig('/tmp/psi.png', dpi=60); plt.close()
````

**Concept Drift vs Covariate Drift & PSI Thresholds (5.5):**

- **Concept drift:** A change in the underlying relationship between input features and the target variable (P(Y|X) changes). For example, customer buying behaviour shifts due to an economic recession, so the same features no longer predict purchases as well.
- **Covariate drift:** A change in the distribution of input features (P(X) changes) while the relationship P(Y|X) stays the same. For example, an age-based model trained on one demographic receives more young users over time.
- **PSI < 0.1:** No significant shift — the distribution is stable; no action required.
- **PSI 0.1–0.2:** Minor shift — monitor closely, investigate the change, but retraining may not yet be needed.
- **PSI > 0.2:** Major shift — the distribution has changed significantly; model retraining or recalibration is recommended.
- **Retraining trigger example:** Schedule an automated retraining pipeline whenever PSI exceeds 0.2 for two consecutive monitoring windows.

### 6.1 Core Concepts

A bag contains: 4 red, 3 blue, 3 green marbles.

1. Define the sample space and compute P(red), P(blue), P(green).
2. Two marbles drawn **without replacement**. Compute:
   - P(first=red, second=blue) — **joint probability**
   - P(second=blue | first=red) — **conditional probability**
3. Are the two draws **independent**? Show mathematically.

````python
total = 10
red, blue, green = 4, 3, 3

p_red   = red   / total
p_blue  = blue  / total
p_green = green / total

# Joint: P(red first AND blue second) = P(red) * P(blue | red drawn)
p_red_then_blue = (red / total) * (blue / (total - 1))

# Conditional: P(blue second | red first) = blue remaining / total remaining
p_blue_given_red = blue / (total - 1)

# Independence check: P(B|A) == P(B)?
independent = (p_blue_given_red == p_blue)

print(f"P(red)={p_red:.3f}, P(blue)={p_blue:.3f}, P(green)={p_green:.3f}")
print(f"P(red,blue): {p_red_then_blue:.4f}")
print(f"P(blue|red): {p_blue_given_red:.4f}")
print(f"P(blue):     {p_blue:.4f}")
print(f"Independent? {independent}")
````

### 6.3 Bayes' Theorem

**Scenario — Spam filter:**
- P(Spam) = 0.30 (prior)
- P("free" | Spam) = 0.80 (likelihood)
- P("free" | Not Spam) = 0.05

1. Compute P(Spam | "free") using Bayes' theorem.
2. Implement a simple `naive_bayes_predict(prior_spam, p_word_given_spam, p_word_given_ham)` that returns the posterior P(Spam | word).
3. In the markdown cell, map each term — **prior, likelihood, evidence, posterior** — to the spam example.

````python
p_spam            = 0.30
p_free_given_spam = 0.80
p_free_given_ham  = 0.05
p_ham             = 1 - p_spam

# P(free) — law of total probability
p_free = p_free_given_spam * p_spam + p_free_given_ham * p_ham

# Bayes
p_spam_given_free = (p_free_given_spam * p_spam) / p_free

print(f"P('free'): {p_free:.4f}")
print(f"P(Spam | 'free'): {p_spam_given_free:.4f}")

def naive_bayes_predict(prior_spam, p_word_given_spam, p_word_given_ham):
    p_word = p_word_given_spam * prior_spam + p_word_given_ham * (1 - prior_spam)
    return (p_word_given_spam * prior_spam) / p_word

pred = naive_bayes_predict(0.30, 0.80, 0.05)
assert abs(pred - p_spam_given_free) < 1e-9
print(f"naive_bayes_predict: {pred:.4f}")
print("6.3 passed ✓")
````

**Bayes Term Mapping (6.3):**

| Term | Formula | In this example |
|------|---------|-----------------|
| Prior | P(Spam) | 0.30 — our initial belief that any email is spam before seeing its content |
| Likelihood | P(word \| Spam) | P("free" \| Spam) = 0.80 — how likely a spam email contains the word "free" |
| Evidence | P(word) | P("free") = 0.275 — overall probability of seeing "free" in any email |
| Posterior | P(Spam \| word) | P(Spam \| "free") ≈ 0.873 — updated belief the email is spam after seeing "free" |

### 6.4 Central Limit Theorem

1. Define a population following an **exponential distribution** (λ=1, size=100,000).
2. Repeatedly draw samples of size `n=30` and record their means — do this 5,000 times.
3. Plot the histogram of sample means.
4. Overlay the **theoretical normal distribution** predicted by the CLT: μ = population mean, σ = population_std / sqrt(n).
5. Run a KS test to confirm the distribution of sample means is approximately normal.

````python
np.random.seed(7)
population = np.random.exponential(scale=1.0, size=100_000)

n_samples = 1000   # Note: very large n_samples causes KS test to reject even true normals
sample_size = 30
sample_means = np.array([np.mean(np.random.exponential(scale=1.0, size=sample_size))
                          for _ in range(n_samples)])

pop_mean = population.mean()
pop_std  = population.std()
clt_std  = pop_std / np.sqrt(sample_size)

print(f"Population mean: {pop_mean:.4f}, Population std: {pop_std:.4f}")
print(f"Sample means mean: {sample_means.mean():.4f}, std: {sample_means.std():.4f}")
print(f"CLT predicted std: {clt_std:.4f}")

# Plot
fig, ax = plt.subplots()
ax.hist(sample_means, bins=40, density=True, alpha=0.6, color='steelblue', label='Sample means')
x_range = np.linspace(sample_means.min(), sample_means.max(), 300)
ax.plot(x_range, stats.norm.pdf(x_range, pop_mean, clt_std),
        color='darkorange', lw=2, label=f'Theoretical N({pop_mean:.2f}, {clt_std:.3f})')
ax.legend(); ax.set_title('CLT: Distribution of Sample Means')
plt.tight_layout(); plt.savefig('/tmp/clt.png', dpi=60); plt.close()

# KS test
ks_result = stats.kstest(sample_means, 'norm', args=(pop_mean, clt_std))
print(f"KS test p-value: {ks_result.pvalue:.4f} → Approximately normal? {ks_result.pvalue > 0.05}")
````

**CLT Reflection (6.4):**

The Central Limit Theorem states that, regardless of the shape of the underlying population distribution, the distribution of sample means converges to a normal distribution as sample size grows. This matters profoundly for statistics and ML because it justifies using normal-distribution-based tools (t-tests, confidence intervals, z-scores) even when the raw data is not normally distributed — as long as samples are sufficiently large (typically n ≥ 30). In machine learning, it underpins assumptions in linear regression (normally distributed residuals), supports bootstrapping methods, and explains why many aggregate metrics (like mean prediction error over a test set) tend to be well-behaved and amenable to classical statistical inference.
